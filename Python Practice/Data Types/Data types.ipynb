#**Python Data Types**
A data type is the category of a value that tells Python what kind of data is stored in a variable, so it knows what operations it is allowed to perform on it.

There are different data types :


* String

* Intiger
* Float


* Complex

* Boolean






#**String**

```python
Name="Anso"
Address="Kalanki"
print(Name)
print(Address)
print(type(Name))
```

#**Integer**

```python
Age=32
Phone_No=98000000
print(Age)
print(Phone_No)
print(type(Age))
```

#**Float**

```python
weight=58.8
height=5.6
print(weight)
print(height)
print(type(weight))
```

#C**omplex**

```python
x=1j+2
print(x)
print(type(x))

```

#B**oolean**

```python
is_sign_in=True
is_working=False
print(is_sign_in)
print(is_working)
print(type(is_sign_in))
```
