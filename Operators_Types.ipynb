#Comparison in **if** statement


*   Comparison operator help python to make decision.

#Logical Operator
##Logical operator combine or reverse condition.
* **and** : True only when both condition are true.
* **Or**  : True when at least one condition is true.
* **not** : Reverses True to False and False to True.



image_cfc5d7bb.jpg

```python
age=20
has_id=True
print(age>18 and has_id )
```

```python
age=20
has_id=False
print(age>18 or has_id)
```

```python
age=17
has_id=False
print(age>18 or has_id)
```

```python
has_email=True
has_number=False
print(has_email or has_number)
```

```python
has_email=False
has_number=False
print(has_email or has_number)
```
