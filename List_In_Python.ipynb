#**List in python**
##list is a built-in, mutable, and ordered collection of items wrapped in square brackets [] and separated by commas. It is one of the most flexible data structures in the language because it can hold elements of different data types (such as integers, strings, floats, and even other lists) and allows duplicate values.

#Creating List

```python
Fruits=["apple","banana","cherry"]
print(Fruits)
```

##Positive Indexing

```python
Student=["ram","Shyam","Hari","Gita","Rita"]
print(Student[0])
print(Student[4])
```

Negative Indexing

```python
Teachers=["gopal","mukesh","dambar","Bima","Sima"]
print(Teachers[-1])
print(Teachers[-2])
```

###Create marks list and calculate total, average, heighest, and lowest marks.

```python
marks=[45,55,60,70,80]
total=sum(marks)
average=total/len(marks)
heighest=max(marks)
lowest=min(marks)

print("Average",average)
print("Heighest",heighest)
print("Lowest",lowest)
```

```python
marks=[35,45,55]
total=sum(marks)
average=total/len(marks)
print("Average :",average)
```

#Adding elements
###append().

```python
name=["ram","shyam","hari"]
name.append("rita")
print(name)
```

```python
Items=["apple","banana","mango"]
Items.append("cherry")
print(Items)
```

###Insert().

```python
Name=["mina","tina","rena"]
Name.insert(1,"jena")
print(Name)
```

```python
Name.insert(0,"sabrina")
print(Name)
```

###Remove()

```python
Name.remove('sabrina')
print(Name)
```

#Removing Elements

```python
fruits=["apple","Banana"]
fruits.remove("apple")
print(fruits)
```

###Pop()

```python
Name.pop()
print(Name)
```

```python
Name.pop(1)
print(Name)
```

```python
Name.pop(3)
print(Name)
```

###loop

```python
Subjects=["English","Nepali","Math"]
for Subject in Subjects:
  print(Subject)
```

##Ascending and descending order.


```python
marks=[39,48,36,44,49]
ascending_marks=sorted(marks)
print(ascending_marks)
```

```python
descending_marks=sorted(marks, reverse=True)
print(descending_marks)
```

###Nested list

```python
Student=[
    ["Gita",80],
    ["Nita",50],
    ["meena",60]
]
print(Student[0])
print(Student[1])
```

```python
Items=[
    ["Iphone",50],
    ["Samsung",20],
    ["Sony",10]
]
print(Items[0])
print(Items[1])
```
