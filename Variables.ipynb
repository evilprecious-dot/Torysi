#V**ariables Naming Rules**
##The following rules should be followed while naming variables:

* A variable name must start with a letters or the underscores character.
* A variable name cannot start with a number.

* Spaces are not allowed in variable names.
* Python variable names are case-sensitive.

* Python keywords cannot be used as variable names.





#**Case Sensitives**
In python variables are case-sensitives. Uppercase and lowercase letters are evaluated as different characters.

Example, Name and name are two different variables.

```python
Name="Torysi"
name="Repo"
print(Name)
print(name)
```

#**Remark**
The output is different because Name and name are different variables in python.

#**Multiple Variables Assignment**
Python allows you to assign values to multiple variable in one line.
##Many Values to multiple Variables


```python
x,y,z='orange','apple','mango'
print(x)
print(y)
print(z)
```

#**Finding the type of a variable**
The type() function is used to check the data type of a value stored in a variable.

```python
a=2
b="apple"
c=1.2
d=a,b,c,1
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
x=1
y=2.8
z="3"
print(type(x))
print(type(y))
print(type(z))


```
